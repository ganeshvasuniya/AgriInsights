```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Styling
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
%matplotlib inline
```

## 1. Load Processed Clean Dataset

```python
data_path = os.path.join('..', 'data', 'processed', 'agricultural_yield_cleaned.csv')
df = pd.read_csv(data_path)
print(f"Loaded clean dataset shape: {df.shape}")
df.head()
```

## 2. Descriptive Summary & Distribution Statistics

```python
crop_stats = df.groupby('Crop')['Yield_Tons_Per_Ha'].agg(
    Count='count',
    Mean='mean',
    Std='std',
    Min='min',
    Median='median',
    Max='max',
    IQR=lambda x: x.quantile(0.75) - x.quantile(0.25)
).round(3)
crop_stats
```

## 3. Multivariate Correlation Heatmap

```python
numeric_cols = [
    'Yield_Tons_Per_Ha', 'Annual_Rainfall_mm', 'Seasonal_Rainfall_mm',
    'Temperature_Avg_C', 'Humidity_Pct', 'Soil_pH', 'Soil_Nitrogen_N',
    'Soil_Phosphorus_P', 'Soil_Potassium_K', 'NPK_Total',
    'Fertilizer_Usage_Kg_Per_Ha', 'Irrigation_Coverage_Pct'
]
plt.figure(figsize=(10, 8))
sns.heatmap(df[numeric_cols].corr(), annot=True, fmt='.2f', cmap='Spectral', center=0)
plt.title('Correlation Matrix: Climate, Nutrients, and Agricultural Yield', weight='bold')
plt.show()
```

## 4. Hypothesis Testing: One-Way ANOVA
- **Test 1**: Soil Type influence on Wheat Yield
- **Test 2**: State / Regional infrastructure influence on Rice Yield

```python
# ANOVA Test: Soil Type on Wheat Yield
wheat_samples = [grp['Yield_Tons_Per_Ha'].values for _, grp in df[df['Crop'] == 'Wheat'].groupby('Soil_Type')]
f_soil, p_soil = stats.f_oneway(*wheat_samples)
print(f"[ANOVA: Soil Type vs Wheat Yield] F-statistic: {f_soil:.4f}, p-value: {p_soil:.4e}")

# ANOVA Test: State on Rice Yield
rice_samples = [grp['Yield_Tons_Per_Ha'].values for _, grp in df[df['Crop'] == 'Rice'].groupby('State')]
f_state, p_state = stats.f_oneway(*rice_samples)
print(f"[ANOVA: State vs Rice Yield] F-statistic: {f_state:.4f}, p-value: {p_state:.4e}")
```
